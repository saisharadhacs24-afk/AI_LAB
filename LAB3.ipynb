```python
import heapq

# Initial and Goal states
INITIAL_STATE = ((2, 8, 3), (1, 6, 4), (7, 0, 5))
GOAL_STATE = ((1, 2, 3), (8, 0, 4), (7, 6, 5))

MOVES = [(-1, 0, 'UP'), (1, 0, 'DOWN'), (0, -1, 'LEFT'), (0, 1, 'RIGHT')]

def misplaced_tiles(state):

    count = 0
    for r in range(3):
        for c in range(3):
            val = state[r][c]
            if val != 0 and val != GOAL_STATE[r][c]:
                count += 1
    return count

def get_neighbors(state):

    # Find position of blank space (0)
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                zr, zc = r, c

    neighbors = []
    for dr, dc, move in MOVES:
        nr, nc = zr + dr, zc + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            grid = [list(row) for row in state]
            grid[zr][zc], grid[nr][nc] = grid[nr][nc], grid[zr][zc]
            neighbors.append((tuple(tuple(r) for r in grid), move))
    return neighbors

def solve_a_star_misplaced(initial_state):

    # Priority Queue stores tuples of: (f_score, g_score, current_state, path)
    h_start = misplaced_tiles(initial_state)
    open_set = [(0 + h_start, 0, initial_state, [])]
    visited = set()

    while open_set:
        f, g, state, path = heapq.heappop(open_set)

        if state in visited:
            continue
        visited.add(state)

        # Goal Check
        if state == GOAL_STATE:
            return path, g

        for neighbor, move in get_neighbors(state):
            if neighbor not in visited:
                g_new = g + 1
                h_new = misplaced_tiles(neighbor)
                f_new = g_new + h_new
                heapq.heappush(open_set, (f_new, g_new, neighbor, path + [move]))

    return None, 0

# Run solver
path, cost = solve_a_star_misplaced(INITIAL_STATE)

print("=== A* Search (Misplaced Tiles) ===")
print("Sequence of Moves:", path)
print("Total Moves (Cost g):", cost)
```

```python
import heapq

# Initial and Goal states
INITIAL_STATE = ((2, 8, 3), (1, 6, 4), (7, 0, 5))
GOAL_STATE = ((1, 2, 3), (8, 0, 4), (7, 6, 5))

MOVES = [(-1, 0, 'UP'), (1, 0, 'DOWN'), (0, -1, 'LEFT'), (0, 1, 'RIGHT')]

# Precompute target positions for fast lookup in Manhattan Distance
GOAL_POSITIONS = {}
for r in range(3):
    for c in range(3):
        val = GOAL_STATE[r][c]
        if val != 0:
            GOAL_POSITIONS[val] = (r, c)

def manhattan_distance(state):
    """Calculates h(n): sum of Manhattan distances of all tiles to goal positions."""
    distance = 0
    for r in range(3):
        for c in range(3):
            val = state[r][c]
            if val != 0:
                target_r, target_c = GOAL_POSITIONS[val]
                distance += abs(r - target_r) + abs(c - target_c)
    return distance

def get_neighbors(state):
    """Generates valid next states by sliding adjacent tiles into the blank space."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                zr, zc = r, c

    neighbors = []
    for dr, dc, move in MOVES:
        nr, nc = zr + dr, zc + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            grid = [list(row) for row in state]
            grid[zr][zc], grid[nr][nc] = grid[nr][nc], grid[zr][zc]
            neighbors.append((tuple(tuple(r) for r in grid), move))
    return neighbors

def solve_a_star_manhattan(initial_state):
    """A* Search algorithm using Manhattan Distance heuristic."""
    h_start = manhattan_distance(initial_state)
    # Priority Queue stores tuples of: (f_score, g_score, current_state, path)
    open_set = [(0 + h_start, 0, initial_state, [])]
    visited = set()

    while open_set:
        f, g, state, path = heapq.heappop(open_set)

        if state in visited:
            continue
        visited.add(state)

        # Goal Check
        if state == GOAL_STATE:
            return path, g

        for neighbor, move in get_neighbors(state):
            if neighbor not in visited:
                g_new = g + 1
                h_new = manhattan_distance(neighbor)
                f_new = g_new + h_new
                heapq.heappush(open_set, (f_new, g_new, neighbor, path + [move]))

    return None, 0

# Run solver
path, cost = solve_a_star_manhattan(INITIAL_STATE)

print("=== A* Search (Manhattan Distance) ===")
print("Sequence of Moves:", path)
print("Total Moves (Cost g):", cost)
```
