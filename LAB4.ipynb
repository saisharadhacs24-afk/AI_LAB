```python
from itertools import combinations

def cost(state):
    attacks = 0
    for i, j in combinations(range(4), 2):
        if (state[i] == state[j] or
            abs(state[i] - state[j]) == abs(i - j)):
            attacks += 1
    return attacks

def hill_climbing(state):
    while True:
        neighbors = []

        for i, j in combinations(range(4), 2):
            nxt = list(state)
            nxt[i], nxt[j] = nxt[j], nxt[i]
            neighbors.append((cost(nxt), (i, j), tuple(nxt)))

        best = min(neighbors, key=lambda x: (x[0], x[1]))

        print("Current:", state, "Cost:", cost(state))
        print("Best neighbor:", best[2], "Cost:", best[0])

        if best[0] >= cost(state):
            return state

        state = best[2]

initial_state = (3, 1, 2, 0)
solution = hill_climbing(initial_state)
print("Final state:", solution)
print("Final cost:", cost(solution))
```
